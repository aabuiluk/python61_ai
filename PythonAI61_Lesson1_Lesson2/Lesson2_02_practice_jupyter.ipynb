## Опис датасету (приклад)

**Мета:** передбачити `passed` (склав іспит чи ні).

**Features:** `hours_study`, `sleep_h`, `practice_tasks`  
**Target:** `passed` (0/1) — supervised classification.

Спочатку запускає файл-зразок `Lesson2_ad_exam_pass_sample.csv`, або ваш `Lesson2_01_my_dataset.csv` після `Lesson2_01_practice_mini_dataset.py`.


```python
from pathlib import Path
import csv

candidates = [
    Path("Lesson2_01_my_dataset.csv"),
    Path("Lesson2_ad_exam_pass_sample.csv"),
]
data_path = next((p for p in candidates if p.exists()), None)
assert data_path is not None, "Немає CSV. Спочатку зробіть Lesson2_01 або використайте sample."

with data_path.open(encoding="utf-8") as f:
    rows = list(csv.DictReader(f))

columns = list(rows[0].keys()) if rows else []
print("file:", data_path.name)
print("rows:", len(rows))
print("columns:", columns)
print("first:", rows[0] if rows else None)
```
