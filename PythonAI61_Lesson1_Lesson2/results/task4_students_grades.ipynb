# Оцінки студентів

Міні-звіт по групі: рахуємо середній/макс/мін бал, знаходимо тих, хто нижче 60, і будуємо текстову візуалізацію успішності.


```python
students = [("Alex", 85), ("Maria", 92), ("John", 74), ("Olga", 60), ("Max", 48)]

scores = [score for _, score in students]
average = sum(scores) / len(scores)
maximum = max(scores)
minimum = min(scores)
below_60 = [name for name, score in students if score < 60]

print("average:", average)
print("maximum:", maximum)
print("minimum:", minimum)
print("below_60:", below_60)
print()
print("Text visualization:")
for name, score in students:
    stars = "*" * (score // 10)
    print(f"{name:<10} | {stars}")
```
