# Звіт із продажів

_У цьому розділі ми рахуємо підсумки за простими даними_


```python
sales = [122, 135, 100, 150, 90, 200, 170]

total = sum(sales)
average = total / len(sales)
maximum = max(sales)
minimum = min(sales)

print("total:", total)
print("average:", average)
print("maximum:", maximum)
print("minimum:", minimum)
```
