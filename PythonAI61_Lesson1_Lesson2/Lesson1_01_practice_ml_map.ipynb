# Lesson 1 · Завдання 1. Карта ML-задачі

Оберіть один кейс і заповніть словник нижче.

Нижче — **приклад даних** (кредитний скоринг). Замініть на свій кейс, якщо хочете.


```python
task = {
    "business_goal": "Зменшити збитки від неплатежів без втрати хороших клієнтів",
    "X": ["дохід", "вік_кредитної_історії", "кількість_прострочень", "регіон"],
    "y": "default_90d",  # 1 = не платить, 0 = платить
    "task_type": "supervised / binary classification",
    "metric": "recall для класу default + контроль precision",
    "risks": {
        "false_negative": "дали кредит — отримали збиток",
        "false_positive": "відмовили надійному клієнту — втратили прибуток",
        "ethics": "уникати proxy-дискримінації",
    },
}

print(task["task_type"], "->", task["metric"])
assert task["business_goal"], "Заповніть business_goal"
assert task["X"], "Додайте хоча б одну ознаку в X"
assert task["y"], "Вкажіть y"
print("OK: карта задачі заповнена")
```
