## Ваш опис (приклад)

Користуюсь рекомендаціями YouTube / Spotify-подібного сервісу. Система вирішує, який контент показати далі, щоб збільшити час перегляду. Можливі ознаки X: історія переглядів, час доби, пристрій, схожість із іншими користувачами. Ціль y — клік / догляд до кінця / оцінка. Метрика успіху — CTR або watch time при контролі різноманітності. Головний ризик — «бульбашка фільтрів» і втрата релевантності для нових інтересів.


```python
product = {
    "name": "YouTube Recommendations",
    "business_decision": "Яке відео показати далі, щоб утримати увагу користувача",
    "X": ["історія_переглядів", "час_доби", "пристрій", "схожі_користувачі"],
    "y": "clicked_or_watched",
    "metric": "CTR / watch time (+ diversity constraint)",
    "main_risk": "filter bubble: користувач бачить лише схожий контент",
}

for key in ("name", "business_decision", "y", "metric", "main_risk"):
    assert product[key], f"Заповніть поле: {key}"
assert product["X"], "Додайте можливі ознаки X"
print("OK:", product["name"])
```
