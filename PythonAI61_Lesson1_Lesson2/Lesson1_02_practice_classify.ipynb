# Lesson 1 · Завдання 2. Класифікація сценаріїв

Для кожного сценарію вкажіть тип навчання і коротке обґрунтування.

Нижче — **приклад відповідей**. Можете змінити формулювання, логіка має лишитись коректною.


1. Групування товарів інтернет-магазину за поведінкою покупців без готових міток.
2. Прогноз температури на завтра за історичними вимірами.
3. Робот-пилосос, що отримує +1 за чисту кімнату і -1 за удар об стіну.
4. Визначення мови відгуку: uk / en / ru.


```python
scenarios = [
    ("Сегментація товарів", "unsupervised", "немає y, шукаємо структуру"),
    ("Температура на завтра", "supervised regression", "є числовий y"),
    ("Робот-пилосос", "reinforcement", "навчання через reward"),
    ("Мова відгуку", "supervised classification", "є дискретний y"),
]

for name, kind, why in scenarios:
    assert kind and why, f"Заповніть відповідь для: {name}"
    print(f"{name}: {kind} ({why})")
```
